```python
class Shape:
  def area():
    pass
  def perimeter():
   pass
class Circle(Shape):
 def __init__(self,r):
   self.r=r
 def area(self):
   return 3.14*self.r*self.r
 def perimeter(self):
   return 2*3.14*self.r
class Square(Shape):
  def __init__(self,s):
    self.s=s
  def area(self):
    return self.s*self.s
  def perimeter(self):
    return 4*self.s
class Triangle(Shape):
  def __init__(self,a,b,c,h):
    self.a=a
    self.b=b
    self.c=c
    self.h=h
  def area(self):
    return 0.5*self.b*self.h
  def perimeter(self):
    return self.a+self.b+self.c
c1= Circle(4)
print("Circle Area:", c1.area())
print("Circle Perimeter:", c1.perimeter())
s1= Square(5)
print("Square Area:", s1.area())
print("Square Perimeter:", s1.perimeter())
t1= Triangle(3,4,5,6)
print("Triangle Area:", t1.area())
print("Triangle Perimeter:", t1.perimeter())



```
